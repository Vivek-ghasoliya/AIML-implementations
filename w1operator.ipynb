```python
~12 #beacuse it is a bitwise operator and it gives you the complement of the number. It is equal to -(n+1) where n is the number. So, ~12 = -(12+1) = -13
```

```python
22>>2 # beacause it is a bitwise operator and it gives you the right shift of the number. It is equal to n//2**x where n is the number and x is the number of bits to shift. So, 22>>2 = 22//2**2 = 22//4 = 5
```

```python
22<<2 # beacause it is a bitwise operator and it gives you the left shift of the number. It is equal to n*2**x where n is the number and x is the number of bits to shift. So, 22<<2 = 22*2**2 = 22*4 = 88
```
