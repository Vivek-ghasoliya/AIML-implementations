```python
class car:
    def __init__(self,year,make,model,speed):
        self.__year =year
        self.__make =make
        self.__model =model
        self.__speed =0
        
    def set_speed(self,speed):
        self.__speed = 0 if speed<0 else speed
    
    def get_speed(self):
        return self.__speed
```

```python
shyam = car(2027,"lambhorghini","advantador",370)
```

```python
shyam.set_speed(900)
```

```python
shyam.get_speed()
```

BANK ACCOUNT SYSTEM

```python
class bank_account:
    def __init__(self,balance):
        self.__balance = balance
    def deposit(self,amount):
        self.__balance = self.__balance + amount
    def withdraw(self,amount):
        if self.__balance>=amount:
            self.__balance = self.__balance - amount
            
            return True
        else:
            return False
        
    def get_balance(self):
        return self.__balance
```

```python
obj_bank_acc = bank_account(1000)
```

```python
obj_bank_acc.get_balance()
```

```python
obj_bank_acc.deposit(10000)
```

```python
obj_bank_acc.get_balance()
```

```python
obj_bank_acc.withdraw(7777)
```

```python
obj_bank_acc.get_balance()
```
