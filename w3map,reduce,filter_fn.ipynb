```python
## create a functiom which square every no of the list and return the equared list
```

```python
l=[1,2,3,45,6,7]
```

```python
def sq(a):
    ans =[]
    for i in a:
        t = i*i
        ans.append(t)
    return ans
```

```python
sq(l)
```
