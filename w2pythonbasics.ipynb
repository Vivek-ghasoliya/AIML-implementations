```python
## fortmat()
name = "Vivek"
age = 23
print("My name is {} and my age is {}".format(age,name)) # we can use {} to print the value of variable inside the string.
```

```python
## placeholder
name = "Vivek"
age = 23
print("My name is {myname} and my age is {myage}".format(myname=name,myage=age))
## now here you can see if we now switch the order in format it still print the correct value because we are using placeholder here.
```

```python
product_prize = int(input("Enter the product price:"))
if product_prize>1000:
    print("the price of the product is :{}".format(product_prize*0.8))
elif product_prize>5000:
    print("the price of the product is :{}".format(product_prize*0.7))
else:
    print("the price of the product is :{}".format(product_prize*0.9))
```
