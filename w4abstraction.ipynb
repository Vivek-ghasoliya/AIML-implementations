## Abstraction

Abstraction means hiding the unnecessary implementation details and showing only the essential features to the user.

Think of a car:

You use: Start,Accelerate,Brake,Steering

But you don't need to know exactly how the engine, fuel injection, or braking mechanism works internally.

```python
class BU:
    def student_details(self):
        pass
    def student_marks(self):
        pass
    def student_attendance(self):
        pass
    def student_fees(self):
        pass
    
```

```python
## there is a abc library in python which is used to create abstract classes and methods.
## we can use the abc library to create abstract classes and methods.
```

```python
import abc
class BU:
    @abc.abstractmethod
    def student_details(self):
        pass
    @abc.abstractmethod
    def student_marks(self):
        pass
    @abc.abstractmethod
    def student_attendance(self):
        pass
    @abc.abstractmethod
    def student_fees(self):
        pass
    """this is a abstract class which contains abstract methods.
    we can not create object of this class. we can only create object of the child class 
    which implements all the abstract methods of the parent class.
    
    this is a way to achieve abstraction in python. 
    abstraction is a way to hide the implementation details and show only the functionality to the user.
    
    this is a blueprint for the child class. 
    the child class must implement all the abstract methods of the parent class."""
    
```

```python
class data_science(BU):
    def student_details(self):
        print("student details")
    def student_marks(self):
        print("student marks")
    def student_attendance(self):
        print("student attenddance")
    def student_fees(self):
        print("student fees")
```

```python
obj = data_science()
```
