CONSTRUCTOR 

A constructor is used to initialize an object when you create it.
In Python, you'll usually see the constructor as __init__().

Why do we use a constructor?
1. To initialize object data

```python
class BU:
    def __init__(self,phone_number,emailid,student_id):
        self.phone_number = phone_number
        self.emailid = emailid
        self.student_id = student_id
    def return_studentInfo(self):
        return self.phone_number ,self.emailid ,self.student_id 
```

```python
vivek =BU(9785783014,"s24cseu2076@bennett.edu.in","s24cseu2076")
```

```python
vivek.return_studentInfo()
```

```python
class data_science:
    def syllabus(self):
        print("this is my method for data science syllabus")
```

```python
class web_dev:
    def syllabus(self):
        print("this is my method for web dev syllabus")
```

```python
def class_parcer(class_obj):
    for i in class_obj:
        i.syllabus()
```

```python
obj_data_sc = data_science()
```

```python
obj_web_dev = web_dev()
```

```python
class_obj = [obj_data_sc,obj_web_dev]
```

```python
class_parcer(class_obj)
```
