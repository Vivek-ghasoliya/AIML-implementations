## Decotrator

A decorator in Python is a function that takes another function, wraps it, and changes or extends what that function does when it is called, without changing the original function's code.

A decorator can:

1.Add something before or after the function.

2.Control whether the original function runs.

3.Change the result or behavior of the function.

4.Do all of this without modifying the original function itself.

```python
def test(a,b):
    print("this is start of my function")
    print(a+b)
    print("this is end of my function")
```

```python
test(10,20)
```

```python
def deco(func):
    def inner(a,b):
        print("this is start of my function")
        func(a,b)
        print("this is end of my function")
    return inner
```

```python
def test1(a,b):
    print(a*b)
```

```python
test1(10,20) ## here you can see that the start ans end fn does not get printed because we have not used the decorator here.
```

```python
deco(test1)(10,20)
```

```python
@deco
def test2(a,b):
    print(a-b)
```

```python
test2(100,20)
```

```python
"""now you can see the start and end of the function is printed because we have used the decorator here."""
```

```python
import time
def timer(func):
    def inner(*args,**kwargs):
        start = time.time()
        func(*args,**kwargs)
        end = time.time()
        print("time taken in this functon is ",end-start)
    return inner
```

```python
@timer
def test3(a,b,c):
    if a>b and a>c:
        print("a is greater")
    elif b>a and b>c:
        print("b is greater")
    else:
        print("c is greater")
```

```python
test3(10,40,20)
```
