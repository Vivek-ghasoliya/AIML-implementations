```python
## we are gonna use yield as function which helps us to print data one by one instead of creating all value at once 
```

```python
def fibbo(n):
        a,b =0,1
        for i in range(n):
            yield a
            a,b = b,a+b
```

```python
fibbo(10)
## this here you can see it gives a object not an answer to your question
## if you want your ans you have to use a for loop for this
```

```python
for i in fibbo(10):
    print(i)
```
